# Task 7: Feature engineering

```python
# AgeGroup
Bank_df["AgeGroup"] = pd.cut(
    Bank_df["age"],
    bins=[17, 29, 39, 49, 59, float("inf")],
    labels=[
        "18 to 29",
        "30 to 39",
        "40 to 49",
        "50 to 59",
        "60+"
    ]
)

# BalanceGroup
Bank_df["BalanceGroup"] = pd.cut(
    Bank_df["balance"],
    bins=[float("-inf"), 0, 1000, 5000, 10000, float("inf")],
    labels=[
        "Negative Balance",
        "Low Balance",
        "Moderate Balance",
        "High Balance",
        "Very High Balance"
    ]
)

# HasAnyLoan
Bank_df["HasAnyLoan"] = (
    (Bank_df["housing"] == "yes") |
    (Bank_df["loan"] == "yes")
).map({True: "yes", False: "no"})

# ContactedPreviously
Bank_df["ContactedPreviously"] = (
    (Bank_df["previous"] > 0) |
    (Bank_df["pay_days"] != -1)
).map({True: "yes", False: "no"})

#DepositNumeric
Bank_df["DepositNumber"] = Bank_df["deposit"].map({
    "yes": 1,
    "no": 0
})

# Display selected columns
print(
    Bank_df[
        [
            "age",
            "AgeGroup",
            "balance",
            "BalanceGroup",
            "housing",
            "loan",
            "HasAnyLoan",
            "pay_days",
            "previous",
            "ContactedPreviously",
            "DepositNumber"
        ]
    ]
)
```

# Task 8: Groupby practice

```python
print(Bank_df)
```

```python
"""
Calculate deposit performance by: 
.Groupby = NewName=("column_name", "calculation")

1. Job
For each job calculate:
Number of records
Number of successful deposits
Deposit success rate
Average balance
Median balance

"""
# Make sure calculation columns are numeric
Bank_df[["balance", "DepositNumber"]] = (
    Bank_df[["balance", "DepositNumber"]].apply(pd.to_numeric)
)

# Calculate deposit performance by job
job_groupAnalysis = Bank_df.groupby("job").agg(
    TotalRecords=("job", "count"),
    TotalSuccDepo=("DepositNumber", "sum"),
    DepositSuccRate=("DepositNumber", "mean"),
    AverageBalance=("balance", "mean"),
    MedianBalance=("balance", "median")
)

# Convert deposit success rate to percentage
job_groupAnalysis["DepositSuccRate"] = (
    job_groupAnalysis["DepositSuccRate"] * 100
)

print(job_groupAnalysis)
```

```python
"""
2. Education
Calculate:
Customer count
Deposit count
Deposit rate
"""

Education_analysis = Bank_df.groupby("education").agg(
    CustomerCount = ("education", "count"),
    DepositCount = ("DepositNumber", "sum"),
    DepositRate = ("DepositNumber", "mean")
)

# Convert deposit rate to percentage
Education_analysis["DepositRate"] = (
    Education_analysis["DepositRate"] * 100
)

print(Education_analysis)
```

```python
"""
Marital status 
Calculate deposit rate
"""
MarriageStatus_analysis = Bank_df.groupby("marital").agg(
    DepositRate = ("DepositNumber", "mean")
)
# Convert deposit rate to percentage
MarriageStatus_analysis ["DepositRate"] = (
    MarriageStatus_analysis ["DepositRate"] * 100
)

print(MarriageStatus_analysis)
```

```python
"""
Month
Calculate:
Number of contacts
Successful deposits
Deposit rate

"""

Month_analysis = Bank_df.groupby("month").agg(
    NumberOfContacts = ("contact", "count"),
    SuccessfulDeposit = ("DepositNumber", "sum"),
    DepositRate = ("DepositNumber", "mean")
)

Month_analysis ["DepositRate"] = (
    Month_analysis ["DepositRate"] *100
)


month_order = [
    "jan", "feb", "mar", "apr", "may", "jun",
    "jul", "aug", "sep", "oct", "nov", "dec"
]
Month_analysis = Month_analysis.reindex(month_order)


print(Month_analysis )
```

```python
"""
Previous campaign outcome
Calculate deposit rate by poutcome.
"""

Campaign_analysis = Bank_df.groupby("outcome").agg(
    DepositRate = ("DepositNumber", "mean")
)

Campaign_analysis ["DepositRate"] = (
    Campaign_analysis ["DepositRate"]*100
)

print(Campaign_analysis)
```

```python
"""
Age group
Calculate:
Count
Average balance
Deposit rate
Then sort each table logically.
"""

AgeGroup_analysis = Bank_df.groupby(
    "AgeGroup",
    observed = False
).agg(
    Count = ("age","count"),
    AverageBalance = ("balance", "mean"),
    DepositRate = ("DepositNumber", "mean")
)

AgeGroup_analysis ["DepositRate"] = (
    AgeGroup_analysis ["DepositRate"]*100
)

print(AgeGroup_analysis)
```
